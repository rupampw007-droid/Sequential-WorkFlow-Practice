```python
from langgraph.graph import StateGraph, MessagesState, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",

)
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState)->BlogState:

    # fetch title 
    title = state['title']

    # call llm gen outline
    prompt = f'Generate a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content

    # update state
    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:

    title = state['title']
    outline = state['outline']

    # call llm gen outline
    prompt = f'Write a detailed blog on title - {title} using the following outline - {outline}'

    content = model.invoke(prompt)

    state['content'] = content

    return state
```

```python
graph = StateGraph(BlogState)

# nodes
graph.add_node('create-outline', create_outline)
graph.add_node('create-blog' , create_blog)

# edges
graph.add_edge(START, 'create-outline')
graph.add_edge('create-outline', 'create-blog')
graph.add_edge('create-blog' , END)

workflow = graph.compile()
```

```python
initial_state = {"title" : "gay"}
final_state = workflow.invoke(initial_state)

print(final_state['content'])
```
